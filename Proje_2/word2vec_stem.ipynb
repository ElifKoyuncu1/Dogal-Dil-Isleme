```python
from gensim.models import Word2Vec
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

df = pd.read_csv('stemize_yorumlar.csv')

input_sentence = "origin decid make switch blackberri android great deal research whole android platform good bad could entir readi whole new learn curv brand new os usual bring plan hit ground run profici possibl new phone end first day ran first went liter dozen store hope get x find back order week cours fault make leap x first came wait found total everywher look order one anyway wait wait week week wait final lost patienc partial go give wife bb tour quit interest trade see old curv dire straight sure long last went order initi impress see review incred unfortun sale rep told thing order get sale turn entir fals name dinc sd card x oh also said froyo updat alreadi done phone imo blatant lie order secur sale work regular sale rep verizon store went suppos could forgiv honest mistak manag kind mistak almost certainli intent one bit tongu howev move day decid want bigger screen came back manag inform would like use option x stock realli could tell anoth would come began get discourag funni thing began happen liter moment began serious consid keep dinc began realli like phone coupl week woke found ota froyo updat wait activ initi excit know phone improv would make even better notic better batteri life improv keep mind dinc batteri woe abysm begin improv web load initi figur updat came new widget would spontan start regardless often use task killer close even forc close use manag applic function would routin check auto task killer notic citi id skype other almost alway run use least moment app yet would time infuri read dozen forum use app like auto task killer realli necessari android design manag function use task killer exit internet manual close app use leav rest alon set atk ignor certain app discov phone still work fine notic absolut zero lag time issu phone slow program run quietli background face unless physic open skype citi id tell someth realli use memori absolut anal shut absolut everyth use task killer look around read forum especi android develop see point accept fact dinc forev go mine got call x show rush made trade wake next day mix emot realli grown accustom htc sens ui got ta say overal much polish better around look system x mean like x though two differ phone use variat exact os one happen polish notic screen amaz x bright obvious huge big x pixel dinc sinc one smaller screen densiti greater therefor amol screen incred also look brilliant vibrant one thing hamper dinc seriou way abil read direct sunlight read major problem dismiss rare use phone way thought amaz often boy suck circumst x hand much easier sunlight see read point goe x place notic anoth differ swipe main screen adjac screen respons incred found sever time x take sever second even regist tri go home screen one either side puzzl compar everi way even gig processor lag time oh well use phone week back eclair x small step backward come eventu absolut love swype x come right box realli look dinc beta trial end swype homepag look look even download similar app call slideit work great day end trial version final found swype free notic one thing version x version download incred microphon option time want use finger speak wish text progam absolut freakin amaz btw one swype keyboard x chang keyboard origin dinc order use function bit annoy realli bother nice see includ also kind miss button along bottom dinc along optic trackpad physic button x seem quit bit differ good way incred respons incred miss day oh well sever reason may like one phone anoth also one reason believ android eventu realli eat iphon sale divers htc make android phone motorola samsung name compani place uniqu version android phone better other option other option make platform amaz hate passion still want iphon sol howev get plenti option want android phone least one model everi carrier us depend provid may abl custom phone true abil droid versu iphon pass recent version mistaken reason platform work thrive insist iphon better say honestli matter opinion point option get iphon verizon realli might get one think iphon awesom recept issu notwithstand thing android simpli ca iphon app market platform huge realli matter phone get app ever need good one develop iphon counterpart android point essenti moot will accept absolut horrif servic guest terribl experi servic bb bold particular area pretti much sour good howev may get fantast servic live matter make updat come day week experi continu grow also feel compel note biggest problem android thu far amaz much fun play serious eat product work let keep mind come back soon see els felt import add review oh thank read updat batteri life x definit better incred say roughli better consid massiv screen say notic enorm differ switch incred x brows internet guess realli know appreci much differ better experi whole froyo use one week switch htc put flash lite inred even without froyo internet realli dramat improv serious quit realiz went surf around yesterday discov world differ iphon user balk flash accord steve job never decreas overal internet view experi delud differ simpli amaz suppos never troll around use phone flash capabl realli realiz miss trust incred set allow view websit mobil phone mode view would desktop comput far found option may download skyfir updat arriv decid realli like abl widget screen wish could whatev size want wo option better none right got ta say realli miss stock htc calendar came incred tri locat anoth version onlin app market compet far come alert light top screen left speaker much easier see ridicul microscop light incred employ difficult see even look almost directli phone face turn slightli one way simpli regist human eye happi improv x one option x dinc abl make alert light blink differ color hey glad see darn thing new option downright x camera par dinc say long decent light photo rock indoor even minor condit photo come amazingli graini see day far either phone camera entir replac regular digit camera far today day fiddl hd video capabl much post updat much hate admit almost almost believ phone big purchas x dinc posit would issu go one quit dramat think inch screen probabl ideal middl ground small larg still love gorgeou crisp color screen x becom use incred size almost feel like tri walk around ipad pocket doubt would issu gone dinc first issu minor best honest factor big one one nevertheless updat impress camera function dx especi comparison camera dinc mostli incred camera plain even gig processor seem take way long camera app load properli deploy moment wish captur may lost enough tri actual snap pic seem take forev shutter eventu snap year old palm treo faster pretti larg margin grant realiz treo could fraction today phone capabl got ta realiz term technolog year ago practic come far thing progress reason interest yesterday read onlin review compar x incred one place dinc beat x camera accord review exact opposit happen person experienc x slow snap pic dinc much faster find bothersom exact phone see total differ result wonder also photo upload facebook much quicker dx dinc even upgrad froyo still waitin updat x btw flipsid speakerphon x abysm incred speaker light year ahead time louder grant use often phone big natur assum speaker would big money best speakerphon ever heard cell phone came old origin blackberri bold great speaker even older bb curv much louder speaker bit surpris one thing neglect motorola mean howev requir loud speaker phone would serious think twice check one person make purchas discov quit accid embarassingli place phone vibrat incom notif may still make audibl recept notif happen yesterday teach lesson church exactli kind place want discov phone entir mode howev found wonder place phone vibrat still happen realli like option howev one place would rather prefer realli simpl put phone silent want entir fuction phone silent much ask sayin updat sale call local sprint store yesterday notic display model htc evo run whim went check updat wait manual check x incred woke show home screen notic updat readi download sure readi go took mayb minut download complet updat nice see middl work day realli ca go without abl use phone updat chang way screen look initi turn small cosmet chang notic calendar function better notic swype screen keyboard look almost differ mostli color chang although felt need alter clue look fine oh well flash run howev notic even though discov tweak internet set order view webpag normal appear desktop instead mobil set simpl fix mind mani websit still render x mobil page version bug crap outta incred issu sinc gone back see origin set save found fact webpag render normal especi flash feel though miss cleaner full internet experi dinc bit frustrat want know whim decid see could download copi movi home comput took bit figur load fairli quickli within minut watch iron man palm hand got ta say pretti dang sweet pictur qualiti awesom render screen beauti well done side note wish moan physic button bottom screen like realiz entir person issu notic mani review incred like button say use x week realli miss design button dinc also much initi fan optic track pad dinc one x highlight problem one problem incred name tri edit anyth written x physic point use fingertip touchscreen phone howev big finger stretch easi get curser place exactli want use fingertip problem optic track pad made ridicul simpl dinc major flaw track pad imo could adjust sensit like could old track ball use older blackberri least could figur consequ make move curser bit slow time infinit accur resort use finger place curser want think extra real estat x screen make much huge differ experi still bit frustrat get blast curser specif letter wish edit may factor mani howev certainli one increasingli annoy own incred prior probabl would even notic wish make inform decis choos x dinc well definit someth consid updat dx dinc first smartphon own without full qwerti keyboard own plenti someth thought long hard buy mostli due relianc addict physic keyboard briefli consid iphon back stuck purgatori ie consid buy one use virtual keyboard look pretti well laid thing consid window mobil tilt month singl worst phone ever own mostli due window mobil htc huge fan landscap keyboard felt actual big use easili thought long hard purchas dinc without keyboard featur simpli awesom made text breez found version swype download sold dx come swype newer version updat download froyo phone month find swype abil miss keyboard fact encourag everyon chanc android base phone get swype never need one ever say lot come someon use feel keyboard absolut essenti text email say still love phone discov new thing new app almost daili android oper system sold thing consid know endors incred x depend person like want extra real estat x screen provid well ca go wrong howev want phone everyth x smaller incred realli pretti well incred say make sure absolut must go store physic handl phone first buy size factor think twice get x fit pocket nuisanc carri everyon feel way homework end pretti satisfi oh good luck updat okay phone nearli year still rock honestli serious love phone time figur thing hope help month notic screen sensit serious make text via swype imposs batteri pull almost everyth without success final googl prob discov mani peopl experienc thing took note one thing miss happen phone plug wall charger soon figur exactli problem issu sinc know screen goe haywir charg long recogn wo eras everyth phone bring back factori spec see exact problem persist boy mistak wipe itch upgrad phone os aka gingerbread quit time supposedli verizon motorola edg releas public offici howev write offici updat releas verizon dx find appear gb updat insist push phone verizon stori deni confirm place whether real anyon guess point found great deal difficulti figur load typic instruct far away head end download comput attach phone drop sd card way phone root btw worri would wipe memori overal updat much cosmet less obviou initi upgrad froyo howev chang good one standpoint task bar top longer white dark blue typic imag whole new look drop task bar also updat one updat dismiss individu prefer small chang sure found conveni much expect person favorit upgrad text write one biggest complaint froyo want find text notic get curser move finger right place challeng time big finger imagin difficult upgrad gb small icon come write text larg enough use point curser wherev want problem whatsoev love new improv along small welcom chang like much improv copi past realli includ begin work also notic improv perform pretti much around virtual lag main screen faster internet faster perform seen sever youtub video chronicl improv detail check leisur say expect gingerbread huge chang cosmet trust chang good one realli help phone mani decid next phone also android see switch anyth appl point droid offer right box awesom devic"

model_files = [
    "stemized_model_cbow_window2_dim100.model",
    "stemized_model_cbow_window2_dim300.model",
    "stemized_model_cbow_window4_dim100.model",
    "stemized_model_cbow_window4_dim300.model",
    "stemized_model_skipgram_window2_dim100.model",
    "stemized_model_skipgram_window2_dim300.model",
    "stemized_model_skipgram_window4_dim100.model",
    "stemized_model_skipgram_window4_dim300.model"
]


def sentence_vector(model, sentence):
    words = str(sentence).split()
    vectors = [model.wv[w] for w in words if w in model.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(model.vector_size)

df_filtered = df[df['content'] != input_sentence].copy()

used_contents = set()

results = []

for model_file in model_files:
    print(f"\nModel: {model_file}")
    model = Word2Vec.load(model_file)

    input_vec = sentence_vector(model, input_sentence)

    df_model = df_filtered[~df_filtered['content'].isin(used_contents)].copy()

    df_model['vector'] = df_model['content'].apply(lambda x: sentence_vector(model, x))

    df_model['similarity'] = df_model['vector'].apply(lambda x: cosine_similarity([input_vec], [x])[0][0])

    top = df_model.sort_values(by='similarity', ascending=False).head(5)

    top5_scores = []
    top5_docs = []
    top5_ids = []
    
    for _, row in top.iterrows():
        content = row['content']
        if content not in used_contents:
            top5_scores.append(row['similarity'])
            top5_docs.append(content)
            top5_ids.append(str(row['document_id']))
            used_contents.add(content)

    avg_score = np.mean(top5_scores) if top5_scores else 0

    results.append({
        "Model Adı": model_file,
        "5 Benzer Metin": ", ".join(top5_docs),
        "5 Benzer Metin ID": ", ".join(top5_ids),
        "Skorlar": ", ".join([f"{score:.4f}" for score in top5_scores]),
        "Ortalama": round(avg_score, 4)
    })

df_results = pd.DataFrame(results)
df_results.to_csv("word2vec_stem.csv", index=False)

print("Sonuçlar 'word2vec_stem.csv' dosyasına kaydedildi.")
```
